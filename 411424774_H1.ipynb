```python
H=float(input("請輸入身高（公分）："))
W=float(input("請輸入體重（公斤）："))
BMI=W/((H*0.01)**2)
if BMI<18.5:
  x="過輕"
else:
  if BMI<24:
    x="正常"
  else:
    x="過重"
print(f"你的 BMI 值為：{BMI:.2f}，判定為：{x}")
```
