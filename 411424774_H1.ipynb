```python
s=7
g=int(input("請猜一個數"))
while g !=s:
  if g>s:
    print("你猜太大了")
  else:
      print("你猜太小了")
  g=int(input("再猜一個數"))
  print("恭喜猜中了")
```

```python
import random
day=1
max_day=3
win=False
while day<=max_day and not win:
  Lucky=random.randint(1,50)
  g=int(input(f"今天第{day}天,請輸入1-50數字"))
  print(f"今天的幸運數字是{Lucky}")
  if g == Lucky:
    print("你中獎了")
    win=True
  else:
    print("你沒中獎，明天再來")
    day+=1
if not win:
  print("很遺憾3天都沒中獎，下次再試試吧！ ")
```
